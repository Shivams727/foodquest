```python
store_file_map = {
    "2025-09-06": "/Volumes/workspace/default/raw/albek-6/",
    "2025-09-07": "/Volumes/workspace/default/raw/albek-7/",
    "2025-09-08": "/Volumes/workspace/default/raw/albek-8/"
}
selected_date = dbutils.widgets.get("selectDate")
```

```python
file_path = store_file_map[selected_date]
print(file_path)
df1 = spark.read.json(
    file_path,
    multiLine=True
)
display(df1)
```

```python
 
from pyspark.sql.functions import *
df_valid = df1.filter(
    (col("isVoid") == False) | (col("isVoid").isNull())
)
# Card payments
df_card = df_valid.select(
    explode(col("payments.cards")).alias("card")
).select(
    
    explode(col("card.detail")).alias("detail"),
    col("card.cardType").alias("cardType"),
    col("detail.bankType").alias("bankType"),
    col("detail.onlineName").alias("onlineName"),
    # col("detail.otherName").alias("otherName"),
    col("detail.amount").cast("double").alias("amount")
).withColumn(
    "PaymentMethod",
    when(col("cardType") == "CreditCard", col("bankType"))
    .when(col("cardType") == "Online", col("onlineName"))
    # .when(col("cardType") == "Other", col("otherName"))
    .otherwise(col("cardType"))
)
# Cash payments
df_cash = df_valid.select(
    
    explode(col("payments.cash")).alias("amount")
).withColumn(
    "PaymentMethod", lit("Cash")
)
# Union card and cash payments
df_all_payments = df_card.select(
     "PaymentMethod", "amount"
).unionByName(
    df_cash.select(
         "PaymentMethod", "amount"
    )
)
daily_sales_payment = df_all_payments.groupBy(
 
    col("PaymentMethod")
).agg(
    sum(col("amount")).alias("Value")
)
display(daily_sales_payment)

df_final = (
    daily_sales_payment.withColumn("date", lit(selected_date))
    .select( "date", *daily_sales_payment.columns
)
)

display(df_final)

 
```
